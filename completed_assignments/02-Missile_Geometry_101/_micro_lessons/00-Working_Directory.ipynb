# 00 — My Working Directory

Before we do anything spatial, we must know:

- What is the current working directory?
- Does the `data/` folder exist?
- If the data doesn't exist, stop nicely without crashing.

We can fix it this time by using `Path.parent` to look one level up if needed.


## Pathlib

`pathlib` documentation: https://docs.python.org/3/library/pathlib.html


```python
from pathlib import Path

# Get path of the current working directory
cwd = Path.cwd()
print("Current Working Directory:")
print(cwd)

# First try: data folder inside current directory
data_path = cwd / "data"
print("\nLooking for data folder at:")
print(data_path)

exists = data_path.exists()
print("\nDoes data folder exist?")
print(exists)
```

## Check parent directory if needed

If the `data/` folder is not inside the current working directory, back up one level and check there.


```python
if not exists:
    parent_data_path = cwd.parent / "data"
    print("\nTrying parent directory instead:")
    print(parent_data_path)

    if parent_data_path.exists():
        data_path = parent_data_path
        exists = True

print("\nFinal data folder path:")
print(data_path)

print("\nDoes data folder exist now?")
print(exists)
```

## Stop nicely if missing

This version does **not** crash with `assert`.  
Instead, it prints a clear message and lets you continue reading the notebook.


```python
if not exists:
    print(
        f"\n❌ STOP: 'data/' folder was not found.\n"
        f"Tried: {cwd / 'data'}\n"
        f"and:   {cwd.parent / 'data'}\n\n"
        "You may need to create the folder or copy the required files into your local project."
    )
else:
    print("\n✅ Environment looks correct.")
```
