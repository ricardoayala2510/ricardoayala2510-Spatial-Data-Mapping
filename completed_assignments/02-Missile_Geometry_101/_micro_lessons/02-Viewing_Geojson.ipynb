# 02 — Viewing GeoJSON

## Goal
Load a GeoJSON object and display it on a map layer.

This version includes the required edits:
- changed the polygon styling
- added a line from MSU to a restaurant
- added a marker over the restaurant

> If the map does not render, use **File → Trust Notebook** in Jupyter.


## Notes

A GeoJSON **FeatureCollection** stores a list of features under the `features` key.

Each feature can have:
- `geometry` → the shape
- `properties` → labels and style info

This notebook uses:
- a **Point** for Bolin Hall
- a **Polygon** for the bounding box
- a **LineString** for the route
- a **Point** for the restaurant


```python
import folium
import json
```

```python
# Modified GeoJSON object
gj = {
    "type": "FeatureCollection",
    "features": [
        {
            "type": "Feature",
            "properties": {
                "name": "Bolin Hall",
                "marker-color": "#00aa00",
                "marker-size": "large",
                "marker-symbol": "college",
                "fill": "#C392FF",
                "fill-opacity": 0.25
            },
            "geometry": {
                "type": "Point",
                "coordinates": [-98.51947, 33.87372]
            },
            "id": 0
        },
        {
            "type": "Feature",
            "properties": {
                "name": "Modified bounding box",
                "fill": "#00c2ff",
                "fill-opacity": 0.15,
                "stroke": "#ff6600",
                "stroke-width": 3,
                "stroke-opacity": 0.9
            },
            "geometry": {
                "type": "Polygon",
                "coordinates": [
                    [
                        [-98.52054301247594, 33.873957083293334],
                        [-98.52054301247594, 33.87010225735909],
                        [-98.51805044396598, 33.87010225735909],
                        [-98.51805044396598, 33.873957083293334],
                        [-98.52054301247594, 33.873957083293334]
                    ]
                ]
            },
            "id": 1
        },
        {
            "type": "Feature",
            "properties": {
                "name": "Restaurant",
                "marker-color": "#ff0000",
                "marker-size": "large",
                "marker-symbol": "restaurant"
            },
            "geometry": {
                "type": "Point",
                "coordinates": [-98.52260, 33.87560]
            },
            "id": 2
        },
        {
            "type": "Feature",
            "properties": {
                "name": "Route from MSU to Restaurant",
                "stroke": "#8a2be2",
                "stroke-width": 4,
                "stroke-opacity": 0.9
            },
            "geometry": {
                "type": "LineString",
                "coordinates": [
                    [-98.51947, 33.87372],
                    [-98.52020, 33.87410],
                    [-98.52110, 33.87470],
                    [-98.52260, 33.87560]
                ]
            },
            "id": 3
        }
    ]
}

gj_type = gj.get("type")
features = gj.get("features", [])

print("GeoJSON type:", gj_type)
print("Feature count:", len(features))

assert gj_type == "FeatureCollection", f"❌ Expected FeatureCollection, got: {gj_type}"
assert len(features) > 0, "❌ FeatureCollection has zero features."

print(json.dumps(gj, indent=2))
```

```python
# Style function for polygons and lines
def style_function(feature):
    props = feature.get("properties", {})
    return {
        "fillColor": props.get("fill", "#3388ff"),
        "color": props.get("stroke", "#3388ff"),
        "weight": props.get("stroke-width", 2),
        "opacity": props.get("stroke-opacity", 1.0),
        "fillOpacity": props.get("fill-opacity", 0.2)
    }

# Build map
m = folium.Map(location=[33.873957083293334, -98.52054301247594], zoom_start=16)

# Add GeoJSON layer
folium.GeoJson(
    gj,
    name="msu_features",
    style_function=style_function,
    tooltip=folium.GeoJsonTooltip(fields=["name"], aliases=["Feature:"])
).add_to(m)

# Add visible markers for the two point features
folium.Marker(
    location=[33.87372, -98.51947],
    tooltip="Bolin Hall",
    popup="Bolin Hall",
    icon=folium.Icon(color="green", icon="info-sign")
).add_to(m)

folium.Marker(
    location=[33.87560, -98.52260],
    tooltip="Restaurant",
    popup="Restaurant",
    icon=folium.Icon(color="red", icon="cutlery", prefix="fa")
).add_to(m)

folium.LayerControl().add_to(m)

m
```

## What changed?

- The polygon now has:
  - cyan fill
  - orange border
  - lower fill opacity
  - thicker outline

- A new **restaurant point** was added.

- A new **LineString** was added from Bolin Hall to the restaurant.

You can keep editing the coordinates or styling values if you want a different look.
