In this notebook, you'll:

- Use `.iloc[]` for position-based selection
- Create boolean masks to filter rows
- Select specific columns and rows

This version uses a small built-in CSV created inside the notebook so it runs without needing `seaborn`.

## 🔎 Select rows by label with `.loc[]`

```python
# .loc[row_label, column_label]
print(df.loc[0])
```

```python
# Row range (inclusive) with selected columns
df.loc[0:4, ["age", "fare", "sex"]]
```

## 🔢 Select rows by position with `.iloc[]`

```python
# iloc uses integer positions
df.iloc[0]
```

```python
# Row slice, column slice
df.iloc[1:4, 0:5]
```

## 🎯 Boolean Masking (Filtering)

```python
# Filter passengers under age 10
kids = df[df["age"] < 10]
kids[["age", "sex", "class"]].head()
```

## 🧠 Compound Conditions

```python
# Adults in First Class who paid over $100
rich_folks = df[(df["age"] > 18) & (df["class"] == "First") & (df["fare"] > 100)]
rich_folks[["age", "class", "fare"]].head()
```

## 🧹 Select Columns Only

```python
df[["sex", "age", "embarked"]].head()
```

## 🎯 Your Turn

- Select the first row using `.iloc[]`, then again using `.loc[]`
- Slice the first 5 rows and select only `survived`, `sex`, and `age`
- Filter the dataset for female passengers over 40
- Create a DataFrame of passengers in 2nd class under age 15
- Use `.loc[]` to select the rows 10–20 and only the columns `fare`, `class`, and `age`

### 🎯 Bonus
Find passengers with missing age and print only their `name`, `age`, and `pclass`


```python
# 1. First row using iloc and loc
print("First row with iloc:")
print(df.iloc[0])

print("\nFirst row with loc:")
print(df.loc[0])

# 2. First 5 rows with selected columns
print("\nFirst 5 rows with survived, sex, and age:")
print(df.loc[0:4, ["survived", "sex", "age"]])

# 3. Female passengers over 40
female_over_40 = df[(df["sex"] == "female") & (df["age"] > 40)]
print("\nFemale passengers over 40:")
print(female_over_40)

# 4. Passengers in 2nd class under age 15
second_class_kids = df[(df["class"] == "Second") & (df["age"] < 15)]
print("\n2nd class passengers under age 15:")
print(second_class_kids)

# 5. Rows 10–20 and only fare, class, and age
print("\nRows 10 to 20 with fare, class, and age:")
print(df.loc[10:20, ["fare", "class", "age"]])

# Bonus: missing age
missing_age = df[df["age"].isnull()][["name", "age", "pclass"]]
print("\nPassengers with missing age:")
print(missing_age)
```
