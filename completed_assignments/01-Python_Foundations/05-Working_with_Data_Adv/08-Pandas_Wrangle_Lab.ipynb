## 📥 1. Load the Dataset

```python
import pandas as pd
from pathlib import Path

data_path = Path("../../data/students.csv")

if data_path.exists():
    df = pd.read_csv(data_path)
else:
    # Fallback sample dataset so the notebook still runs
    df = pd.DataFrame({
        "first_name": ["Danielle", "Angel", "Joshua", "Jeffrey", "Jill", None, "Bridget", "Craig"],
        "last_name": ["Wood", "Clark", "Adams", "Zuniga", "Wong", "Lynch", "Johnson", None],
        "math_score": [100, 67, 61, 77, 75, 74, None, 94],
        "science_score": [95, 78, 100, None, 83, 96, 100, 96]
    })

df.head()
```

## 🔎 2. Inspect and Audit the Data

```python
# Basic overview
df.info()
```

```python
# Summary stats
df.describe()
```

## 🧼 3. Clean Missing or Invalid Data

```python
# Check for missing values
df.isnull().sum()
```

```python
# Drop rows with missing names
df = df.dropna(subset=["first_name", "last_name"])

# Fill any missing scores with column average
df["math_score"] = df["math_score"].fillna(df["math_score"].mean())
df["science_score"] = df["science_score"].fillna(df["science_score"].mean())

df
```

## 🧠 4. Feature Engineering (New Columns)

```python
# Add average and grade
df["average_score"] = (df["math_score"] + df["science_score"]) / 2

def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "D"

df["grade"] = df["average_score"].apply(grade)
df.head()
```

## 🔽 5. Sorting and Filtering

```python
# Top performers
df[df["average_score"] > 90].sort_values(by="average_score", ascending=False).head()
```

## 📊 6. Group and Describe by Grade

```python
# How many of each grade?
df["grade"].value_counts()
```

```python
# Average scores per grade group
df.groupby("grade")[["math_score", "science_score", "average_score"]].mean()
```

## 💾 7. Save the Cleaned Dataset

```python
df.to_csv("student_scores_cleaned.csv", index=False)
```

## 🧠 Challenge Task

```python
# Filter out students who got a D, sort by last name, and export
d_students = df[df["grade"] == "D"][["first_name", "last_name", "grade"]].sort_values(by="last_name")
d_students.to_csv("d_students.csv", index=False)

d_students
```
