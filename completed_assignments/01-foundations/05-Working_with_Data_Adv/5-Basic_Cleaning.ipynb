# 🧼 Cleaning Dirty Data (Missing Values & Type Fixes)

## Learning Goals

- Detect and count missing values (`NaN`)
- Fill or drop missing data
- Convert column data types safely
- Understand the difference between `NaN`, `None`, `""`, and type mismatches


## 🧪 1. Load a Messy Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None],
    "Age": ["25", "thirty", 35, np.nan, "40"],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022"],
    "Score": [95.5, None, 88.0, 92.5, ""]
}

df = pd.DataFrame(data)
df
```

## 🧯 2. Detecting Missing or Broken Values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

Notice that an empty string `""` is not always counted as `NaN` automatically.  
That matters for columns like `Score`, where the blank value may still need cleaning.


## 🧹 3. Cleaning Strategy Options

```python
df.fillna({
    "Name": "Unknown",
    "Age": -1,
    "Signup Date": "1970-01-01",
    "Score": 0.0
})
```

```python
df.dropna()
```

## 🧬 4. Data Type Fixes

```python
df.dtypes
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df["Signup Date"] = pd.to_datetime(df["Signup Date"], errors="coerce")
df.dtypes
```

Using `errors="coerce"` turns bad values like `"thirty"` or `"not a date"` into `NaN`/`NaT`, which makes them easier to fix.


## 🩹 5. Impute (Fill In) Fixed Missing Values

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Score"] = df["Score"].fillna(df["Score"].mean())
df["Signup Date"] = df["Signup Date"].fillna(df["Signup Date"].min())
df["Name"] = df["Name"].fillna("Unknown")

df
```

## 🤓 6. Cleaned Data Review

```python
df.info()
```

```python
df.describe(include="all")
```

```python
# Example practice version
practice_data = {
    "Name": ["Ana", "", None, "Leo", "??"],
    "Age": ["21", None, "old", "19", ""],
    "Signup Date": ["2023-01-01", "", "tomorrow", None, "2022-12-15"],
    "Score": [91, "", None, "88.5", "bad"],
    "City": ["Dallas", None, "", "Austin", "123"]
}

practice_df = pd.DataFrame(practice_data)
practice_df
```

```python
# One possible cleaning pass for the practice data
practice_df = practice_df.replace("", np.nan)

practice_df["Age"] = pd.to_numeric(practice_df["Age"], errors="coerce")
practice_df["Score"] = pd.to_numeric(practice_df["Score"], errors="coerce")
practice_df["Signup Date"] = pd.to_datetime(practice_df["Signup Date"], errors="coerce")

practice_df["Name"] = practice_df["Name"].fillna("Unknown")
practice_df["City"] = practice_df["City"].fillna("Unknown")
practice_df["Age"] = practice_df["Age"].fillna(practice_df["Age"].median())
practice_df["Score"] = practice_df["Score"].fillna(practice_df["Score"].mean())
practice_df["Signup Date"] = practice_df["Signup Date"].fillna(practice_df["Signup Date"].min())

practice_df
```
