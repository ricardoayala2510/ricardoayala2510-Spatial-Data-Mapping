## 2. Writing and Reading Text Files

```python
# Writing to a text file
with open("example.txt", "w") as f:
    f.write("Hello, Jupyter!\n")
    f.write("This is a text file.\n")
```

```python
# Reading from a text file
with open("example.txt", "r") as f:
    contents = f.read()

contents
```

## 3. JSON Files

```python
import json

data = {"name": "Graylian", "type": "mascot", "skills": ["howl", "parrot talk"]}

# Write JSON
with open("data.json", "w") as f:
    json.dump(data, f)

# Read JSON
with open("data.json", "r") as f:
    loaded = json.load(f)

loaded
```

### Your Turn

Create a dictionary with your name and 2-3 favorite courses. Save it to `courses.json`, then read it back.


```python
courses = {
    "name": "Alisson",
    "favorite_courses": ["Python", "Data Science", "Machine Learning"]
}

with open("courses.json", "w") as f:
    json.dump(courses, f)

with open("courses.json", "r") as f:
    loaded_courses = json.load(f)

loaded_courses
```

## 4. CSV Files

```python
import csv

# Writing CSV
rows = [["Name", "Score"], ["Alice", 90], ["Bob", 85]]
with open("scores.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)

# Reading CSV
with open("scores.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

### Your Turn

Create a CSV with 3 rows of your own data (for example, hobbies). Read it back and display the rows.


```python
hobby_rows = [
    ["Hobby", "Hours per Week"],
    ["Reading", 5],
    ["Gym", 4],
    ["Movies", 3]
]

with open("hobbies.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(hobby_rows)

with open("hobbies.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## Summary

- Use `with open(...)` to safely read and write files.
- JSON is useful for structured data with key-value pairs.
- CSV is useful for tabular data with rows and columns.
