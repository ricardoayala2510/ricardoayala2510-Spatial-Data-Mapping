## How to Install Our Own Lib

Our project contains reusable Python code that lives in:

```text
src/wdo
```

Instead of placing Python modules directly in the project root, we use a **`src` layout** to keep the code organized and separate from:

- notebooks
- datasets
- markdown files
- setup scripts

### Example structure

```text
.
├── data
├── notebooks
├── pyproject.toml
└── src
    └── wdo
```

Notice the `pyproject.toml` file. This is important.


## What `pyproject.toml` Does

The `pyproject.toml` file tells Python’s packaging tools:

1. the project name
2. the Python version requirements
3. where the package code lives (`src/`)

So Python knows that the real code is here:

```text
src/wdo
```


## What This Command Does

Before running notebooks or scripts, run:

```bash
pip install -e .
```

This command:

1. installs the project into the current virtual environment
2. makes the `wdo` package importable anywhere in the project
3. uses **editable mode**, so changes to the source code are immediately available


## Why Editable Mode Matters

After running:

```bash
pip install -e .
```

we can write:

```python
from wdo.geo import haversine_km
```

and Python will find the code inside:

```text
src/wdo
```

Even better, if we modify files in `src/wdo`, we do **not** need to reinstall the package.


## Quick Summary

Run this once after you clone or copy the files into your project:

```bash
pip install -e .
```

When it works, you should usually see something like:

```text
src
├── wdo
└── wdo.egg-info
```

Without this step, Python may not know where to find the `wdo` package.

## Problem One

- Import `haversine_km`
- Calculate the distance between:

```text
[25.7853, -80.188189]
[34.046211, -118.256405]
```

- What is the distance in miles?
- What cities are we working with?

These coordinates are approximately:

- **Miami, Florida**
- **Los Angeles, California**


````python
from wdo.geo import haversine_km

lat1, lon1 = 25.7853, -80.188189
lat2, lon2 = 34.046211, -118.256405

distance_km = haversine_km(lat1, lon1, lat2, lon2)
distance_miles = distance_km * 0.621371

print("Distance (km):", distance_km)
print("Distance (miles):", distance_miles)
print("Cities: Miami, Florida  →  Los Angeles, California")
````

### Expected approximate answer

If your local `wdo.geo.haversine_km` matches the standard haversine formula, the result should be about:

- **3759.19 km**
- **2335.63 miles**


## Problem Two

Calculate a bounding box around the same two city points using `bbox_latlon`.


````python
from wdo.geo import bbox_latlon

lat1, lon1 = 25.7853, -80.188189
lat2, lon2 = 34.046211, -118.256405

# Depending on your local library version, one of these may be the correct call.
try:
    bbox = bbox_latlon(lat1, lon1, lat2, lon2)
except TypeError:
    bbox = bbox_latlon([(lat1, lon1), (lat2, lon2)])

print("Bounding box:")
print(bbox)
````

### Reference min/max values

A simple min/max bounding box around the two points would be:

- south = **25.7853**
- west = **-118.256405**
- north = **34.046211**
- east = **-80.188189**

Your library may return the same values in a tuple, dict, or helper object.


## Problem Three

Using GeoJSON.io as a helper:

- create a polygon approximately the size of a small state
- create one point inside the polygon
- create one point outside the polygon
- run `point_in_polygon` to verify the results


````python
# A simple rectangle near north Texas / Oklahoma area
# You can also draw your own in geojson.io and replace these coordinates.

from wdo.geo import point_in_polygon

# Try to import LatLon if your local package exposes it
try:
    from wdo.geo import LatLon

    polygon = [
        LatLon(33.0, -100.5),
        LatLon(35.0, -100.5),
        LatLon(35.0, -98.0),
        LatLon(33.0, -98.0),
    ]

    inside_point = LatLon(34.0, -99.2)
    outside_point = LatLon(36.0, -99.2)

    print("Inside point result:", point_in_polygon(inside_point, polygon))
    print("Outside point result:", point_in_polygon(outside_point, polygon))

except ImportError:
    # If LatLon is not available in your local version, try plain tuples
    polygon = [
        (33.0, -100.5),
        (35.0, -100.5),
        (35.0, -98.0),
        (33.0, -98.0),
    ]

    inside_point = (34.0, -99.2)
    outside_point = (36.0, -99.2)

    print("Inside point result:", point_in_polygon(inside_point, polygon))
    print("Outside point result:", point_in_polygon(outside_point, polygon))
````

### What you should see

If everything is working correctly, the output should be roughly:

```text
Inside point result: True
Outside point result: False
```
