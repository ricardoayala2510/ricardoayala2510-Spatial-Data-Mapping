# 🧮 Sorting & Filtering in Pandas

## Learning Goals

- Sort rows using `.sort_values()`
- Filter data using conditional logic
- Chain multiple filters together
- Use `.query()` for readable filtering


## 📥 1. Load the Dataset

```python
import pandas as pd
from pathlib import Path

data_path = Path("../../data/students.csv")

if data_path.exists():
    df = pd.read_csv(data_path)
else:
    sample_data = {
        "first_name": ["Danielle", "Angel", "Joshua", "Jeffrey", "Jill", "Erica", "Bridget", "Craig"],
        "last_name": ["Wood", "Clark", "Adams", "Zuniga", "Wong", "Lynch", "Johnson", "Ferrell"],
        "math_score": [100, 67, 61, 77, 75, 74, 92, 94],
        "science_score": [95, 78, 100, 99, 83, 96, 100, 96]
    }
    df = pd.DataFrame(sample_data)

df.head()
```

## 🔢 2. Sorting Data

```python
# Sort by math_score
df.sort_values(by="math_score")
```

```python
# Sort by science_score descending
df.sort_values(by="science_score", ascending=False)
```

```python
# Sort by multiple columns
df.sort_values(by=["science_score", "math_score"], ascending=[False, True])
```

## 🧪 3. Filtering with Conditions

```python
# Who scored at least 90 in math?
df[df["math_score"] >= 90]
```

```python
# Who scored less than 70 in either subject?
df[(df["math_score"] < 70) | (df["science_score"] < 70)]
```

## 🔗 4. Combining Filters

```python
# Students with B+ or higher in both
df[(df["math_score"] >= 87) & (df["science_score"] >= 87)]
```

## 🧾 5. Using `.query()` for Readability

```python
# Same query using .query()
df.query("math_score >= 87 and science_score >= 87")
```

## 🧪 Try It Yourself

- Sort the dataset by first name alphabetically
- Find students with the same math and science score
- Use `.query()` to find students who scored over 95 in either subject


```python
# Sort by first name alphabetically
df.sort_values(by="first_name")
```

```python
# Students with the same math and science score
df[df["math_score"] == df["science_score"]]
```

```python
# Students who scored over 95 in either subject
df.query("math_score > 95 or science_score > 95")
```

## ✅ Demo Mini-Challenge

This small demo gives you something runnable even if `data/survey.csv` is not available.


```python
survey = pd.DataFrame({
    "name": ["Ana", "Luis", "Mia", "Noah", "Sara"],
    "signup_date": ["2024-04-01", "2024-05-10", None, "2024-02-20", "N/A"],
    "completed": ["Yes", "No", "Yes", None, "Yes"],
    "score": [88, 72, 95, 60, 91]
})

survey["signup_date"] = pd.to_datetime(survey["signup_date"], errors="coerce")

# Sort by most recent signup date
survey_sorted = survey.sort_values(by="signup_date", ascending=False)
survey_sorted
```

```python
# Filter out incomplete rows
survey_clean = survey_sorted.dropna(subset=["completed"])
survey_clean = survey_clean[survey_clean["completed"] != "N/A"]
survey_clean
```

```python
# Use query() with a chosen condition
survey_clean.query("score >= 90")
```
