## 🧺 Lists

`list`: Ordered, mutable, and great for storing collections of items.


```python
fruits = ["apple", "banana", "cherry", "dragonfruit"]
print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])

fruits.append("elderberry")
print("Updated list:", fruits)
```

## 🔐 Tuples

`tuple`: Like lists, but immutable. They are useful for fixed data.


```python
coordinates = (34.05, -118.25)  # Latitude, Longitude
print("Coordinates:", coordinates)

# Uncomment this to see the error
# coordinates[0] = 0
```

## 📦 Dictionaries

`dict`: Key-value pairs for fast lookup.


```python
student = {
    "name": "Alice",
    "GPA": 3.9,
    "major": "Data Science"
}

print("Student name:", student["name"])
print("GPA using .get():", student.get("GPA"))

student["minor"] = "Philosophy"
print("Updated dictionary:", student)
```

## 🎯 Sets

`set`: Unordered collections of unique values.


```python
courses = {"CS101", "CS102", "CS103", "CS101"}  # Duplicate ignored
print("Courses:", courses)

print("Is CS102 in the set?", "CS102" in courses)

courses.add("CS104")
courses.discard("CS101")
print("Modified courses:", courses)
```
