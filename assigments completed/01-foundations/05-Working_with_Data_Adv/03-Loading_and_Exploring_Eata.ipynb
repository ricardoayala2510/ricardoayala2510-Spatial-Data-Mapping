# 📂 Notebook 03: Loading and Exploring Data

Welcome to the real world — where data lives in messy CSVs and your job is to make sense of it.

In this notebook, you’ll:

- Load a CSV file using `pd.read_csv()`
- Use `.head()`, `.tail()`, `.info()`, and `.describe()` to explore your data
- Identify potential issues like missing values and odd data types

This version avoids `seaborn` so it can run even if your environment has plotting/style issues.


```python
import pandas as pd
```

## 📥 Load a CSV

Replace the filename below with a real CSV path if you have one.  
For a guaranteed working demo, this notebook creates a small CSV first and then loads it with `pd.read_csv()`.


```python
sample_data = """name,age,gender,fare,survived,cabin
Alice,23,female,72.5,1,C85
Bob,35,male,8.05,0,
Charlie,29,male,13.0,1,E12
Diana,41,female,26.55,1,
Evan,19,male,7.75,0,
Fatima,33,female,53.1,1,B20
Grace,,female,15.5,1,
Henry,50,male,31.0,0,C101
"""

with open("passengers.csv", "w") as f:
    f.write(sample_data)

df = pd.read_csv("passengers.csv")
df.head()
```

## 📊 Peek at the Data

```python
print("First 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())
```

## 🧠 Understand Structure

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
```

## 🧼 Data Types & Missing Values

```python
print("\nInfo:")
df.info()

print("\nMissing values per column:")
print(df.isnull().sum())
```

## 📈 Summary Statistics

```python
df.describe(include="all")
```

## ✏️ Renaming Columns

```python
df.rename(columns={"gender": "sex_label"}, inplace=True)
df.head(2)
```

- Load a dataset of your choice using `pd.read_csv()`
- Print the first and last 5 rows
- Show `.info()` and `.describe()` results

For practice below, we use another small CSV created in the notebook.


```python
my_data = """student,major,gpa,hours_studied,passed
Ana,CS,3.8,12,True
Luis,Math,3.1,8,True
Mia,Biology,3.9,14,True
Noah,History,2.6,5,False
Sara,Physics,,10,True
"""

with open("students_explore.csv", "w") as f:
    f.write(my_data)

my_df = pd.read_csv("students_explore.csv")

print("First 5 rows:")
print(my_df.head())

print("\nLast 5 rows:")
print(my_df.tail())

print("\nColumn names:")
print(my_df.columns.tolist())

print("\nInfo:")
my_df.info()

print("\nDescribe:")
print(my_df.describe(include="all"))

my_df.rename(columns={"hours_studied": "study_hours"}, inplace=True)

print("\nAfter renaming a column:")
print(my_df.head())
```

## 🎯 Bonus

```python
percent_missing_rows = df.isnull().any(axis=1).mean() * 100
print(f"Percent of rows with any missing values: {percent_missing_rows:.2f}%")
```
