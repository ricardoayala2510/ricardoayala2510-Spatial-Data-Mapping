# 🧱 Column Operations: Creating, Renaming, and Dropping Columns

## Learning Goals

- Create new columns using calculations or functions
- Rename one or multiple columns
- Drop unwanted columns safely
- Apply functions across columns using `.apply()` and lambdas


## 🏗️ 1. Setup and Sample Data

```python
import pandas as pd

data = {
    "first_name": ["Alice", "Bob", "Charlie", "David"],
    "last_name": ["Smith", "Jones", "Brown", "Wilson"],
    "math_score": [85, 90, 78, 92],
    "science_score": [88, 85, 82, 95]
}

df = pd.DataFrame(data)
df
```

## ➕ 2. Creating New Columns

```python
# Simple column arithmetic
df["average_score"] = (df["math_score"] + df["science_score"]) / 2
df
```

```python
# Create full name column
df["full_name"] = df["first_name"] + " " + df["last_name"]
df
```

## ✍️ 3. Renaming Columns

```python
# Rename selected columns
df = df.rename(columns={"math_score": "Math", "science_score": "Science"})
df
```

```python
# Rename all columns to uppercase
df.columns = [col.upper() for col in df.columns]
df
```

## ❌ 4. Dropping Columns

```python
# Drop by name
df = df.drop(columns=["FULL_NAME"])
df
```

## 🔁 5. Applying Functions Across Columns

```python
# Categorize based on average score
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    else:
        return "C"

df["GRADE"] = df["AVERAGE_SCORE"].apply(grade)
df
```

## ✅ Worked Mini-Challenge Demo

This demo creates a small sample dataset so you can practice the exact steps even if `data/students.csv` is not available.


```python
students = pd.DataFrame({
    "Student Name": ["Ana", "Luis", "Mia"],
    "Test 1": [85, 90, 88],
    "Test 2": [92, 84, 91],
    "Final Exam": [89, 87, 94],
    "Empty Column": [None, None, None]
})

students
```

```python
# Rename columns with spaces
students.columns = [col.strip().replace(" ", "_") for col in students.columns]

# Create total_score from numeric columns
numeric_cols = students.select_dtypes(include="number").columns
students["total_score"] = students[numeric_cols].sum(axis=1)

# Drop all-NaN columns
students = students.dropna(axis=1, how="all")

students
```

## 📝 Summary

| Action | Method |
|---|---|
| Create column | `df["new"] = ...` |
| Rename column(s) | `df.rename(columns={...})` |
| Rename all columns | `df.columns = [...]` |
| Drop column(s) | `df.drop(columns=[...])` |
| Apply function | `df["col"].apply(func)` |
