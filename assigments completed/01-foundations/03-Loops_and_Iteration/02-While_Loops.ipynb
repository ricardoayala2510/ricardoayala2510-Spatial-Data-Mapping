# 02 - While Loops

Learn how to repeat logic using conditions.

## Learning Goals
- Use `while` with a condition
- Understand infinite loops and loop control


## Count up to 5

```python
counter = 1
while counter <= 5:
    print('Count:', counter)
    counter += 1
```

## Avoid infinite loop

```python
x = 0
while x < 3:
    print('x is', x)
    x += 1
```

## Break and continue

```python
n = 0
while True:
    n += 1
    if n == 3:
        continue
    if n == 6:
        break
    print(n)
```
