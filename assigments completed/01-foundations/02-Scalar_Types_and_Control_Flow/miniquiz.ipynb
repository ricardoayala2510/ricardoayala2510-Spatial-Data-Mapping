## 1️⃣ Which type is returned by this expression?

`type(True)` returns `bool`.

```python
type(True)  # Answer: bool
```

## 2️⃣ Write a boolean statement that would print true if the value in `z` was between `x` and `y`.

```python
x = 7
y = 15
z = 10

print(x < z < y)  # Answer 2
```

## 3️⃣ Add one keyword to your boolean statement from above that would print false instead of true.

Add `not` in front of the statement.

```python
x = 7
y = 15
z = 10

print(not (x < z < y))  # Answer 3
```

## 4️⃣ Fill in the blank to check for inequality

Use `==` so the message prints only when the score is exactly 100.

```python
score = 100

if score == 100:
    print("Perfect score!")
```

## 5️⃣ Add to the `if` statement so that it will also print `"Variable is odd"` depending on the number.

Use an `else` block.

```python
a = 11

if a % 2 == 0:
    print("Variable is even")
else:
    print("Variable is odd")  # Answer 5
```
