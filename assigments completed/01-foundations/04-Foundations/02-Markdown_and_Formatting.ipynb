## 5. Mixing Code and Markdown

```python
# Example: create data in Python
numbers = [1, 2, 3, 4, 5]
squares = [n**2 for n in numbers]
squares
```

Now we explain results in Markdown:

- The list `numbers` contained 1–5.
- The list `squares` contains each number squared.
- Mixing narrative + results is why notebooks are so powerful.


### Your Turn

Create a list of 3 numbers, calculate their cubes in code, then explain in Markdown.


```python
nums = [2, 4, 6]
cubes = [n**3 for n in nums]
cubes
```

The list `nums` contains the values 2, 4, and 6.

The list `cubes` contains each of those values raised to the third power:
- 2³ = 8
- 4³ = 64
- 6³ = 216

This shows how code can generate results and Markdown can explain what they mean.
