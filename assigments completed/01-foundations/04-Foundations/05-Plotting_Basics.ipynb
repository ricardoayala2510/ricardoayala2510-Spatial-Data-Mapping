```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Line Plot

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)")
plt.title("Line Plot Example")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.show()
```

### Your Turn

Plot `cos(x)` on the same graph with a different label.


```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")
plt.title("sin(x) and cos(x)")
plt.xlabel("x")
plt.ylabel("value")
plt.legend()
plt.show()
```

## 3. Scatter Plot

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = 300 * np.random.rand(50)
colors = np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, alpha=0.6, cmap='viridis')
plt.title("Scatter Plot Example")
plt.xlabel("x")
plt.ylabel("y")
plt.colorbar(label="color scale")
plt.show()
```

### Your Turn

Generate two sets of random points and plot them on the same scatter plot.


```python
x1 = np.random.rand(30)
y1 = np.random.rand(30)

x2 = np.random.rand(30)
y2 = np.random.rand(30)

plt.scatter(x1, y1, alpha=0.7, label="Group 1")
plt.scatter(x2, y2, alpha=0.7, label="Group 2")
plt.title("Two Random Point Groups")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

## 4. Bar Plot

```python
categories = ["A", "B", "C", "D"]
values = [5, 7, 3, 8]

plt.bar(categories, values)
plt.title("Bar Plot Example")
plt.xlabel("Category")
plt.ylabel("Value")
plt.show()
```

## 5. Customizing Plots

```python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label="sin(x)", linestyle="--", color="blue")
plt.plot(x, y2, label="cos(x)", linestyle=":", color="red")
plt.title("Customized Line Plot")
plt.xlabel("x (radians)")
plt.ylabel("value")
plt.legend()
plt.grid(True)
plt.show()
```

### Your Turn

Customize your own plot with labels, legend, and grid lines.


```python
x = np.linspace(0, 10, 100)
y1 = x
y2 = x**2

plt.plot(x, y1, label="y = x", linestyle="--")
plt.plot(x, y2, label="y = x^2", linestyle=":")
plt.title("Customized Plot Example")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

## Summary

- `%matplotlib inline` makes plots display inside the notebook.
- Line, scatter, and bar plots are the most common basics.
- Titles, labels, legends, and grids improve readability.
- Visuals tell stories about data more effectively than raw numbers.
