## Q4. File I/O

Write Python code that saves `Hello World` to a file called `hello.txt`, then reads it back and prints the contents.

```python
with open("hello.txt", "w") as file:
    file.write("Hello World")

with open("hello.txt", "r") as file:
    contents = file.read()

print(contents)
```

## Q6. Plotting

Create a line plot for values from 0 to 10.

**Note:** The formula in the prompt appears to be missing, so this answer uses **y = x²** as a reasonable example.


```python
%pip install matplotlib
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11)
y = x ** 2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y = x^2")
plt.title("Line Plot of y = x^2")
plt.show()
```

## Q9. Timing

Compare the runtime of squaring numbers using a Python loop vs. NumPy. Use `%timeit` for each.

```python
import numpy as np

nums = list(range(100000))
arr = np.arange(100000)
```

```python
%timeit [n ** 2 for n in nums]
```

```python
%timeit arr ** 2
```
