# 08 - Timing and Performance

In data science, performance matters. Some code runs fast, some slow.  
Jupyter/IPython gives us tools to measure runtime easily.

In this notebook you will learn:

- `%time` and `%timeit` for single expressions
- `%%time` and `%%timeit` for entire cells
- Comparing loops, list comprehensions, and NumPy
- Why performance awareness is important


## 1. `%time`

```python
%time sum(range(1_000_000))
```

### Your Turn

Use `%time` to measure how long it takes to sort a list of 1 million random numbers.


```python
import random

big_list = [random.random() for _ in range(1_000_000)]
%time sorted(big_list)
```

## 2. `%timeit`

```python
numbers = list(range(1_000))
%timeit [x**2 for x in numbers]
```

### Your Turn

Compare `%timeit` results for a list comprehension vs. a `for` loop that builds the same list.


```python
%timeit [x**2 for x in numbers]
```

```python
%timeit
result = []
for x in numbers:
    result.append(x**2)
```

## 3. `%%time` for a Whole Cell

```python
%%timeit
total = 0
for i in range(1_000_000):
    total += i
total
```

### Your Turn

Wrap a longer multi-line operation with `%%time` to measure its runtime.


```python
%%time
values = []
for i in range(500_000):
    values.append(i * 2)
sum(values)
```

## 4. Comparing Loops vs. NumPy

```python
import numpy as np

numbers = np.arange(1_000_000)

# Python list comprehension
%timeit [x**2 for x in numbers]

# NumPy vectorized
%timeit numbers**2
```

### Your Turn

Try squaring numbers with a Python loop, list comprehension, and NumPy array. Compare times.


```python
python_numbers = list(range(1_000_000))
numpy_numbers = np.arange(1_000_000)
```

```python
%timeit
result = []
for x in python_numbers:
    result.append(x**2)
```

```python
%timeit [x**2 for x in python_numbers]
```

```python
%timeit numpy_numbers**2
```

## 5. Why This Matters

- Performance differences become huge with large datasets.
- Vectorized operations like NumPy and Pandas are usually faster.
- `%timeit` is useful when deciding how to implement something.


## Summary

- `%time` and `%timeit` measure execution speed.
- `%%time` and `%%timeit` work on whole cells.
- Loops are slower than list comprehensions, which are often slower than NumPy.
- Always measure performance before optimizing.
