# 📈 Summary Statistics

Summary statistics help us describe a dataset with just a few numbers. Instead of staring at thousands of rows, we can quickly understand center, spread, and shape.

This notebook covers:

- Measures of central tendency (mean, median, mode)
- Measures of spread (range, variance, standard deviation, IQR)
- Skewness and kurtosis
- Quick summaries with Pandas


## 1. Central Tendency

These describe the “middle” of the data:

- **Mean** – average  
- **Median** – middle value  
- **Mode** – most frequent value


```python
%pip install scipy
```

```python
import numpy as np
import pandas as pd
from scipy import stats

data = [5, 7, 8, 5, 10, 12, 7, 7, 6, 9]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data, keepdims=True)[0][0])
```

### Question

**Which is more robust to outliers, the mean or the median?**

**Answer:** The **median** is more robust to outliers because extreme values affect it much less than they affect the mean.


## 2. Spread

Spread tells us how variable the data are.

- **Range**: max – min  
- **Variance**: average squared deviation from the mean  
- **Standard Deviation**: square root of variance  
- **Interquartile Range (IQR)**: middle 50% (`Q3 - Q1`)


```python
print("Range:", np.max(data) - np.min(data))
print("Variance:", np.var(data, ddof=1))
print("Standard Deviation:", np.std(data, ddof=1))
print("IQR:", stats.iqr(data))
```

### Exercise: Add an outlier

Add an extreme outlier like `100` and compare the mean, median, and standard deviation.


```python
data_with_outlier = data + [100]

print("Original mean:", np.mean(data))
print("New mean:", np.mean(data_with_outlier))
print()

print("Original median:", np.median(data))
print("New median:", np.median(data_with_outlier))
print()

print("Original standard deviation:", np.std(data, ddof=1))
print("New standard deviation:", np.std(data_with_outlier, ddof=1))
```

## 3. Shape: Skewness & Kurtosis

- **Skewness** measures asymmetry
- **Kurtosis** measures how heavy the tails are or how prone the data are to extreme values


```python
print("Skewness:", stats.skew(data))
print("Kurtosis:", stats.kurtosis(data))
```

### Note

High kurtosis suggests heavier tails and more extreme outliers.  
Low kurtosis suggests a flatter distribution.


## 4. Quick Summaries with Pandas

Instead of computing each statistic one by one, Pandas can summarize numeric data quickly with `.describe()`.


```python
df = pd.DataFrame({"Values": data})
df.describe()
```
