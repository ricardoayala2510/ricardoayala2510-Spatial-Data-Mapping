## Goal

- Most projects keep notebooks, scripts, and datasets in different folders.
- You should learn how to build paths to those files safely so your code files and data files will be correct and organized.


## Example Project Layout

```text
project/
│

└── output/
```

If the notebook runs from `project/`, then the file path to `countries.geojson` is:

```python
Path("data/countries.geojson")
```


````python
from pathlib import Path

data_file = Path("data/countries.geojson")

print("Path being checked:", data_file)
print("Exists:", data_file.exists())
````

## Why Check Existence First?

Checking whether a file exists helps students debug path mistakes before trying to load the file.

That means fewer crashes and fewer emotional speeches directed at the laptop.


````python
possible_files = [
    Path("data/countries.geojson"),
    Path("data/missing.geojson"),
    Path("output"),
]

for p in possible_files:
    print(f"{p!s:<28} exists={p.exists()}  is_file={p.is_file()}  is_dir={p.is_dir()}")
````

## Exercise A

In your own words, describe the difference between these three methods. For each one, state what it returns when given: a path to a real file, a path to a real directory, and a path to something that doesn't exist at all.

- `.exists()`
- `.is_file()`
- `.is_dir()`

### Answer

- `p.exists()` checks whether *anything* is present at that path.  
  - Real file: `True`  
  - Real directory: `True`  
  - Missing path: `False`

- `p.is_file()` checks whether the path points to a regular file.  
  - Real file: `True`  
  - Real directory: `False`  
  - Missing path: `False`

- `p.is_dir()` checks whether the path points to a directory.  
  - Real file: `False`  
  - Real directory: `True`  
  - Missing path: `False`

## Optional Advanced — Set Up a Local Workspace

Use `pathlib` (no shell commands) to:

1. Create a `data/` folder inside `00-Paths/` using `Path.mkdir(exist_ok=True)`
2. Create an `output/` folder the same way
3. Copy one json file from the module's data folder into your new local `data/` using `shutil.copy2`
4. Re-run the existence check from above and confirm all three paths are found

Hint: `Path.mkdir(parents=True, exist_ok=True)` will not error if the folder already exists.


````python
from pathlib import Path
import shutil

# Create a local data/ and output/ folder inside the current folder if they don't exist
local_root = Path.cwd()
local_data = local_root / "data"
local_output = local_root / "output"

local_data.mkdir(parents=True, exist_ok=True)
local_output.mkdir(parents=True, exist_ok=True)

# Find one JSON or GeoJSON file in the module's data folder (one level up)
module_data = local_root.parent / "data"

source_file = None
for pattern in ("*.json", "*.geojson"):
    matches = list(module_data.glob(pattern))
    if matches:
        source_file = matches[0]
        break

if source_file is None:
    print("❌ No .json or .geojson file was found in the module data folder:", module_data)
else:
    destination = local_data / source_file.name
    shutil.copy2(source_file, destination)
    print("Copied:", source_file)
    print("To    :", destination)

    # Re-run existence checks
    check_paths = [
        module_data,
        local_data,
        destination,
        local_output,
    ]

    print("\nExistence checks:")
    for p in check_paths:
        print(f"{p!s:<50} exists={p.exists()}  is_file={p.is_file()}  is_dir={p.is_dir()}")
````
