## 1) Using `range`
Using `range`, create a list that starts at 1, ends at 10 (inclusive), increments by 2. Store it in `nums` and print it.

```
nums = list(range(1, 11, 2))
print(nums)
```

## 2) Skipping an Iteration
Iterate 0 through 6. When value is 3, skip it. Print the rest.

```
for i in range(0, 7):
    if i == 3:
        continue
    print(i)
```

## 3) Stopping a Loop Early
Iterate 1 through 10. Stop entirely when 6 is reached. Print each number before stopping.

```
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

## 4) Loop Output
Rewrite the loop exactly and run it (do not change the logic).

```
for i in range(3):
    print("Loop:", i)
```

## 5) File Iteration with `with`
Open `data.txt` for reading using `with`, iterate over each line, and print it.

```
with open("data.txt", "r") as f:
    for line in f:
        print(line, end="")
```
