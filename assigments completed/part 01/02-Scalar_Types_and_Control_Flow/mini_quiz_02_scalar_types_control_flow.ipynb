## 2) Boolean statement: print True if z is between x and y

Given:
```python
x = 7
y = 15
# z = ???
```

````
x = 7
y = 15
z = 10  # example value

# Answer 2:
print(x < z < y)  # True if z is strictly between x and y
````

## 3) Add one keyword to make it print False instead of True

(Use the same statement as #2, but add one keyword.)

````
# Answer 3:
print(not (x < z < y))  # 'not' flips True -> False and False -> True
````

## 4) Fill in the blank to check for inequality

```python
if score ___ 100:
    print("Perfect score!")
```

````
score = 100

# Answer 4: use != for inequality
if score != 100:
    print("Perfect score!")
else:
    print("Not perfect.")
````
