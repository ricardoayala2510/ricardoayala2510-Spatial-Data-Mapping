# Mini Quiz — Python Data Containers
This notebook contains the answers (with runnable code).

## 1) Remove `31` from a list by value
Given:
```python
L = [13, 17, 31, 37, 73]
```
One line that removes `31` **by value**:

````
L = [13, 17, 31, 37, 73]
L.remove(31)  # removes by value (not by index)
L
````

> Note: After removing 31, the list becomes `[13, 17, 37, 73]`.

## 2) Access the 2nd element of a tuple
Given:
```python
T = (13, 17, 31, 37, 73)
```
One line that stores the second element in `x`:

````
T = (13, 17, 31, 37, 73)
x = T[1]  # second element (index 1)
x
````

## 3) Change the 2nd element of a tuple to 11
Tuples are **immutable**, so you can't change an element in place. The usual way is to create a new tuple:

````
T = (13, 17, 31, 37, 73)
T = (T[0], 11, *T[2:])  # create a new tuple with the updated value
T
````

## 4) Print the keys of a dictionary
Given:
```python
D = {
    "apple": "10",
    "banana": "11",
    "cherry": "8"
}
```
Code that prints the keys:

````
D = {
    "apple": "10",
    "banana": "11",
    "cherry": "8"
}

print(D.keys())          # prints a view of keys
# Or, one per line:
for k in D:
    print(k)
````
