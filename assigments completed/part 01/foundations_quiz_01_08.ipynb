## Q4. File I/O
Save `Hello World` to `hello.txt`, then read it back and print it.

```
# Write to the file
with open("hello.txt", "w") as f:
    f.write("Hello World")

# Read from the file and print
with open("hello.txt", "r") as f:
    contents = f.read()

print(contents)
```

## Q6. Plotting
Create a line plot of $y = x^2$ for $x$ values from 0 to 10.

```
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 11)   # 0 to 10 inclusive
y = x**2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y = x^2")
plt.title("Line plot of y = x^2 (x = 0..10)")
plt.show()
```

## Q9. Timing
Compare the runtime of squaring numbers using a Python loop vs. NumPy. Use `%timeit` for each.

```
import numpy as np

nums_list = list(range(1_000_000))
nums_np = np.array(nums_list)

# Python loop
%timeit [n*n for n in nums_list]

# NumPy
%timeit nums_np * nums_np
```
