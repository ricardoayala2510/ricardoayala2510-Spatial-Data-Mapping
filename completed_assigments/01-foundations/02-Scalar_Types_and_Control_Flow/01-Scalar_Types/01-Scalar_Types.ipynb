# 01 - Scalar Types

Review and practice working with Python's basic scalar data types.

## Learning Goals
- Recognize scalar types (`int`, `float`, `str`, `bool`)
- Use the `type()` function to identify types
- Use variables and assignment

## Define some variables

```python
num = 42
pi = 3.14159
name = 'Python'
is_cool = True

print(num)
print(pi)
print(name)
print(is_cool)
```

## Check their types

```python
print(type(num))
print(type(pi))
print(type(name))
print(type(is_cool))
```

## Comparison Practice

Use comparison operators to check values.

```python
x = 10
y = 5
print(x > y)
print(x == y)
print(x != y)
```

## Quick Check

What will the following print?

```python
print(3.0 == 3)
```
